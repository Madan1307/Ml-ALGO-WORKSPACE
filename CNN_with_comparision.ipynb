```python
from sklearn. linear_model import Perceptron
from sklearn.metrics import accuracy_score

from tensorflow.keras.models import Sequential
from tensorflow.keras. layers import Dense
from tensorflow.keras.layers import Conv2D
from tensorflow. keras. layers import Flatten

from tensorflow.keras. layers import MaxPooling2D
from tensorflow.keras.layers import Dropout

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
from sklearn.preprocessing import LabelEncoder , StandardScaler
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical
```

```python
from tensorflow.keras.datasets import mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
x_train
```

```python
y_train
```

```python
x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
x_train[0]
```

using a simple perceptron


```python
y_train_cat=to_categorical(y_train,10)
y_test_cat=to_categorical(y_test,10)
```

```python
perceptron = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(10, activation="softmax")
])
```

```python
perceptron.compile(
    optimizer="sgd",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

perceptron.fit(x_train, y_train_cat, epochs=10,validation_data=(x_test,y_test_cat))
```

```python
perceptron.evaluate(x_test,y_test_cat)[1]
```

Using ANN


```python
ann = Sequential([
Flatten(input_shape=(28,28)),
Dense(128, activation="relu"),
Dense(64, activation="relu"),
Dense(10, activation="softmax")])
```

```python
ann.compile(
    optimizer="sgd",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
ann.fit(x_train,y_train_cat, epochs=10,validation_data=(x_test,y_test_cat))
```

```python
ann.evaluate(x_test,y_test_cat)[1]
```

CNN

```python
x_train_cnn =x_train.reshape(-1,28,28,1)
x_test_cnn =x_test.reshape(-1,28,28,1)
```

```python
model = Sequential([
    Conv2D(32, kernel_size=(3, 3), activation="relu", input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),

    Conv2D(64, kernel_size=(3, 3), activation="relu"),
    MaxPooling2D((2, 2)),

    Flatten(),

    Dense(128, activation="relu"),

    Dense(10, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.fit(
    x_train_cnn,
    y_train_cat,
    epochs=5,
    batch_size=16,
    validation_data=(x_test_cnn, y_test_cat)
)
```

```python
model.evaluate(x_test_cnn,y_test_cat)[1]
```

```python
import matplotlib.pyplot as plt
import numpy as np
import random

# Select 5 random unique indices from the test set
random_indices = random.sample(range(len(x_test)), 5)

# Prepare inputs only for the selected random indices
x_test_sample = x_test[random_indices]
x_test_cnn_sample = x_test_cnn[random_indices]

# Predictions for these samples
perceptron_pred = np.argmax(
    perceptron.predict(x_test_sample, verbose=0),
    axis=1
)

ann_pred = np.argmax(
    ann.predict(x_test_sample, verbose=0),
    axis=1
)

cnn_pred = np.argmax(
    model.predict(x_test_cnn_sample, verbose=0),
    axis=1
)

# Show 5 random images
plt.figure(figsize=(15, 5))

for i, idx in enumerate(random_indices):
    plt.subplot(1, 5, i + 1)

    plt.imshow(x_test[idx], cmap="gray")
    plt.axis("off")

    plt.title(
        f"True: {y_test[idx]}\n"
        f"Perceptron: {perceptron_pred[i]}\n"
        f"ANN: {ann_pred[i]}\n"
        f"CNN: {cnn_pred[i]}"
    )

plt.tight_layout()
plt.show()
```
