```python
import scipy.cluster.hierarchy as sc
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_iris
df=load_iris()
df=pd.DataFrame(df.data,columns=df.feature_names)
df
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaled_data=scaler.fit_transform(df)
```

```python
scaled_data
```

```python
from IPython.core.interactiveshell import no_op
from sklearn.decomposition import PCA
pca=PCA(n_components=2)
x_scaled=pca.fit_transform(scaled_data)
```

```python
x_scaled
```

```python

plt.scatter(x_scaled[:,0],x_scaled[:,1])
```

```python
from sklearn.cluster import AgglomerativeClustering
fig=plt.figure(figsize=(10,8))
dendrogram=sc.dendrogram(sc.linkage(x_scaled,'ward'))
plt.show()
```

```python
hier=AgglomerativeClustering(n_clusters=3,metric="euclidean",linkage="ward")
labels = hier.fit_predict(x_scaled)
```

```python
plt.scatter(x_scaled[:,0],x_scaled[:,1],c=labels)
```

```python
from sklearn.metrics import silhouette_score
score=silhouette_score(x_scaled,labels)
```

```python
from sklearn.datasets import make_moons
x,y=make_moons(n_samples=500,random_state=42,noise=0.05)
```

```python
plt.scatter(x[:,0],x[:,1])
```

```python
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
```

```python
dbscan=DBSCAN(eps=0.5)
labels=dbscan.fit_predict(x_scaled)
```

```python
plt.scatter(x_scaled[:,0],x_scaled[:,1],c=labels)
```

```python
score=silhouette_score(x_scaled,y)
```
