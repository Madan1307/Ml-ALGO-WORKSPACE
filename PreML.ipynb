## **Feature Engineering and Extraction**

```python
sns.histplot(df_clean['bmi'])
```

```python
df_clean["bmi_category"]=pd.cut(
    df_clean['bmi'],
    bins=[0,18.5,24.9,29.9,100],
    labels=["underweight","healthy","overweight","obese"]
)
```

```python
df_clean
```

```python
df_clean=pd.get_dummies(df_clean,columns=['bmi_category'],dtype=int)
```

```python
df_clean
```

```python
df_clean
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scale_cols = ['age', 'bmi', 'children']
df_clean[scale_cols] = scaler.fit_transform(df_clean[scale_cols])
```

```python
df_clean
```

```python
df_clean.columns
```

```python
selected_features=['age', 'is_female', 'bmi', 'children', 'is_smoker', 'charges',
       'region_northeast', 'region_northwest', 'region_southeast',
       'region_southwest', 'bmi_category_underweight', 'bmi_category_healthy',
       'bmi_category_overweight', 'bmi_category_obese']

from scipy.stats import pearsonr
correlations = {
feature: pearsonr(df_clean[feature], df_clean['charges'])[0]
for feature in selected_features
}
```

```python
co_relation_df=pd.DataFrame(list(correlations.items()),columns=['feature','correlation'])
```

```python
plt.figure(figsize=(20,20))
sns.heatmap(df_clean.corr(numeric_only=True),annot=True)
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

df_clean['charges_bin'] = pd. qcut(df_clean['charges'], q=4, labels=False)
chi2_results = {}

for col in selected_features:
  contingency = pd.crosstab(df_clean[col], df_clean['charges_bin'])
  chi2_stat, p_val,_, _= chi2_contingency(contingency)
  decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'
  chi2_results [col] = {
    'chi2_statistic': chi2_stat,
    'p_value': p_val,
    'Decision': decision
  }

chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by='p_value')
chi2_df
```
